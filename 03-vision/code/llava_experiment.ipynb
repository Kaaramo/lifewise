```python
import ollama
response = ollama.chat(
    model="llava:7b",
    messages=[
        {
            "role": "user",
            "content": "Describe this image. What objects can you see?",
            "images": ["../sample test/sample1.jpg"],
        }
    ],
)

print(response["message"]["content"])
```

```python
import ollama
import time

start = time.perf_counter()

stream = ollama.chat(
    model="llava:7b",
    messages=[{
        "role": "user",
        "content": "Briefly describe this image.",
        "images": ["../sample test/sample1.jpg"],
    }],
    stream=True,
    options={"num_predict": 100},
)

for chunk in stream:
    print(chunk["message"]["content"], end="", flush=True)

print(f"\nTime: {time.perf_counter() - start:.1f} seconds")
```
